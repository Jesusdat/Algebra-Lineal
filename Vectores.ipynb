```python
import numpy as np

# Vector de desplazamiento o dirección 3D (X = Este, Y = Norte, Z = Vertical/Profundidad)
# Supongamos un vector de movimiento: 3 cm/año al Este, 4 cm/año al Norte, y baja 2 cm/año en profundidad
v = np.array([3.0, 4.0, -2.0])

# 1. Calcular la magnitud total del vector (Velocidad total)
magnitud = np.linalg.norm(v)

# 2. Vector unitario (dirección puramente orientada)
v_unitario = v / magnitud

# 3. Calcular el ángulo de inclinación/plunge/dip respecto a la horizontal
# Usamos la componente vertical (|v_z|) y la magnitud total
angulo_rad = np.arcsin(abs(v[2]) / magnitud)
angulo_deg = np.degrees(angulo_rad)

print(f"Vector de entrada [X, Y, Z]: {v}")
print(f"Magnitud total: {magnitud:.2f} cm/año")
print(f"Vector unitario: {v_unitario}")
print(f"Ángulo de inclinación (dip/plunge): {angulo_deg:.2f}°")
```

```python
# Dos vectores de dirección (normal a la falla y dirección del esfuerzo)
v1 = np.array([1, 0, 0])
v2 = np.array([0.707, 0.707, 0])

# Producto punto
cos_theta = np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
angulo_entre_vectores = np.degrees(np.arccos(cos_theta))

print(f"Ángulo entre las dos direcciones: {angulo_entre_vectores:.2f}°")
```

```python
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])

# Suma y resta de vectores
suma = u + v        # [5, 7, 9]
resta = u - v       # [-3, -3, -3]

# Multiplicación por un escalar (escalar-vector)
escalar = 3 * u     # [3, 6, 9]

# Multiplicación elemento a elemento (Hadamard)
prod_elem = u * v   # [4, 10, 18]

# Mostrar los resultados en pantalla
print("Vector u:", u)
print("Vector v:", v)
print("Suma (u + v):", suma)
print("Resta (u - v):", resta)
print("Escalar (3 * u):", escalar)
print("Multiplicación elemento a elemento (u * v):", prod_elem)
```

```python
import numpy as np

# Supongamos dos vectores trazados dentro del plano de una falla:
# Vector v1: Dirección Noreste en la falla (X=Este, Y=Norte, Z=Profundidad/Vertical)
v1 = np.array([10.0, 10.0, -5.0])

# Vector v2: Dirección Sudeste en la falla
v2 = np.array([10.0, -5.0, -8.0])

# 1. Producto Cruz para hallar la Normal al plano de falla
vector_normal = np.cross(v1, v2)

# 2. Normalizar el vector normal (obtener el vector unitario)
normal_unitaria = vector_normal / np.linalg.norm(vector_normal)

print("--- PRODUCTO CRUZ EN GEOLOGÍA ESTRUCTURAL ---")
print(f"Vector normal al plano de falla: {vector_normal}")
print(f"Vector normal unitario: {normal_unitaria}")
```

```python
import numpy as np

# Vector del esfuerzo principal mayor (\sigma_1) actuando sobre la región (en MegaPascales - MPa)
sigma_1 = np.array([0.0, 0.0, -50.0])  # Esfuerzo compresional vertical de 50 MPa

# Vector normal unitario a la falla (calculado previamente o derivado del buzamiento)
# Falla inclinada a 45° respecto a la vertical
n_falla = np.array([0.7071, 0.0, 0.7071])

# 1. Producto Punto para calcular la magnitud del esfuerzo normal proyectado
esfuerzo_normal = np.dot(sigma_1, n_falla)

# 2. Calcular el ángulo entre el esfuerzo regional y la falla
cos_theta = np.dot(sigma_1, n_falla) / (np.linalg.norm(sigma_1) * np.linalg.norm(n_falla))
angulo_deg = np.degrees(np.arccos(abs(cos_theta)))

print("\n--- PRODUCTO PUNTO EN GEOMECÁNICA ---")
print(f"Magnitud del esfuerzo normal sobre la falla: {abs(esfuerzo_normal):.2f} MPa")
print(f"Ángulo de impacto entre sigma_1 y la normal de la falla: {angulo_deg:.2f}°")
```
