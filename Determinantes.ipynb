```python
import numpy as np

# Definimos una matriz A
A = np.array([
    [1.0, 0.5],
    [0.0, 0.3]
])

# 1. Calcular el determinante
det_A = np.linalg.det(A)
print(f"El determinante de A es: {det_A:.4f}")

# 2. Buenas prácticas: Validar antes de resolver
# Usamos una tolerancia (1e-9) por los errores de redondeo de la computadora
if abs(det_A) > 1e-9:
    print("La matriz es invertible. Podemos resolver el sistema.")
else:
    print("ALERTA: La matriz es singular (determinante cero o cercano a cero).")
    if abs(np.linalg.det(A)) < 1e-9:
        raise ValueError("¡Error geológico! Los minerales seleccionados son redundantes.")
```
