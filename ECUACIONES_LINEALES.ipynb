```python
import numpy as np

# Sistema de ecuaciones para encontrar la pendiente (gradiente) y el intercepto
# Ecuación 1: 100 * m + b = 20
# Ecuación 2: 500 * m + b = 36

A = np.array([
    [100.0, 1.0],
    [500.0, 1.0]
])

b = np.array([20.0, 36.0])

# Resolver el sistema Ax = b (donde x = [m, b])
m, b_intercept = np.linalg.solve(A, b)

# Calcular la temperatura a 1000 metros de profundidad
profundidad_objetivo = 1000.0
temp_estimada = m * profundidad_objetivo + b_intercept

print(f"Gradiente geotérmico: {m * 1000:.2f} °C por kilómetro")
print(f"Temperatura estimada a {profundidad_objetivo}m: {temp_estimada:.2f} °C")
```

```python
import numpy as np

# Matriz A: Coeficientes de masa y volúmenes inversos
A = np.array([
    [1.0, 1.0],           # Suma de masas = 1000 g
    [1.0/2.7, 1.0/3.0]    # Suma de volúmenes = 356.5 cm3
])

b = np.array([1000.0, 356.5])

# Resolver el sistema
masas = np.linalg.solve(A, b)

print(f"Gramos de Granito: {masas[0]:.2f} g")
print(f"Gramos de Basalto: {masas[1]:.2f} g")
```

```python
import numpy as np

# Sistema de ecuaciones lineales muy sencillo para 3 pozos:
# Ecuación 1: Nivel en P1 fijo = 50
# Ecuación 2: P1 - 2*P2 + P3 = 0 (Equilibrio de Darcy en el pozo central)
# Ecuación 3: Nivel en P3 fijo = 30

A = np.array([
    [1.0,  0.0,  0.0],
    [1.0, -2.0,  1.0],
    [0.0,  0.0,  1.0]
])

b = np.array([50.0, 0.0, 30.0])

# Resolver para obtener las cargas hidráulicas [P1, P2, P3]
cargas = np.linalg.solve(A, b)

print(f"Carga hidráulica en Pozo 1: {cargas[0]:.1f} m")
print(f"Carga hidráulica en Pozo 2 (Central): {cargas[1]:.1f} m")
print(f"Carga hidráulica en Pozo 3: {cargas[2]:.1f} m")
```
